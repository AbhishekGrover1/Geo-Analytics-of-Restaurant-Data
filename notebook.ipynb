## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
print("Libraries loaded.")
```

## 2. Load Dataset

```python
df = pd.read_csv("../Dataset/Dataset.csv")
df["Cuisines"] = df["Cuisines"].fillna("Unknown")
df["Primary Cuisine"] = df["Cuisines"].apply(lambda x: x.split(",")[0].strip())
geo_df = df[(df["Latitude"].abs() > 0.01) & (df["Longitude"].abs() > 0.01)].copy()
print(f"Total: {df.shape[0]} | Valid GPS: {geo_df.shape[0]} | Cities: {df['City'].nunique()}")
print(f"Latitude : {df['Latitude'].min():.2f} to {df['Latitude'].max():.2f}")
print(f"Longitude: {df['Longitude'].min():.2f} to {df['Longitude'].max():.2f}")
```

## 3. Global Distribution Map

```python
fig, ax = plt.subplots(figsize=(14, 7), facecolor="#1a1a2e")
ax.set_facecolor("#16213e")
sc = ax.scatter(geo_df["Longitude"], geo_df["Latitude"],
                c=geo_df["Aggregate rating"], cmap="YlOrRd", s=8, alpha=0.6)
cbar = fig.colorbar(sc, ax=ax, shrink=0.7)
cbar.set_label("Rating", color="white")
cbar.ax.yaxis.set_tick_params(color="white")
ax.set_title("Global Restaurant Distribution", fontsize=14, color="white", fontweight="bold")
ax.set_xlabel("Longitude", color="white")
ax.set_ylabel("Latitude", color="white")
ax.tick_params(colors="white")
plt.tight_layout()
plt.show()
```

## 4. City-Level Analysis

```python
city_stats = df.groupby("City").agg(
    count=("Restaurant ID","count"),
    avg_rating=("Aggregate rating","mean"),
    avg_cost=("Average Cost for two","mean"),
).reset_index().sort_values("count", ascending=False)
print("Top 10 Cities:")
print(city_stats.head(10).to_string(index=False))
```

```python
top20 = city_stats.head(20)
fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(top20["City"], top20["count"], color=plt.cm.tab20.colors, edgecolor="white")
ax.set_title("Top 20 Cities by Restaurant Count", fontweight="bold")
ax.set_xlabel("City")
ax.set_ylabel("Count")
ax.tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.show()
```

## 5. Cuisine Analysis

```python
top10_c = df["Primary Cuisine"].value_counts().head(10)
fig, ax = plt.subplots(figsize=(9, 9))
ax.pie(top10_c.values, labels=top10_c.index, autopct="%1.1f%%",
       colors=plt.cm.Set3.colors[:10], startangle=140)
ax.set_title("Top 10 Primary Cuisine Share", fontweight="bold")
plt.tight_layout()
plt.show()
```

## 6. Business Insights

**1. New Delhi dominates** with 5,473 restaurants but has one of the lower average ratings (2.44) — high saturation, variable quality.  
**2. Smaller cities (Lucknow, Guwahati)** have average ratings above 4.0 — established, quality-driven markets.  
**3. North Indian cuisine** is the most common primary cuisine nationally.
